# Label-level aggregate baselines

Compute group-level score and ranking metrics; no person-level examples are retained.

```
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))
import helpers
```

```
labels_train_biased, labels_train_blind, labels_test_biased, labels_test_blind, _, _, _, _, _, _ = helpers.build_features('none')

gender_test = helpers.load_gender_test()
male = (gender_test == 0)
female = (gender_test == 1)
```

```
blind_KL =helpers.computeKL(labels_test_blind[male], labels_test_blind[female])
blind_mean_male = labels_test_blind[male].mean()
blind_mean_female = labels_test_blind[female].mean()

biased_KL = helpers.computeKL(labels_test_biased[male], labels_test_biased[female])
biased_mean_male = labels_test_biased[male].mean()
biased_mean_female = labels_test_biased[female].mean()


print('Blind:', blind_KL)
print('Mean male (Blind):', blind_mean_male)
print('Mean female (Blind):', blind_mean_female, "\n")
# ----
print('Biased:', biased_KL)
print('Mean male (Biased):', biased_mean_male)
print('Mean female (Biased):', biased_mean_female)
```

```
dp_blind, p_val_blind = helpers.computeDemographicParity(labels_test_blind, gender_test)

dp_biased, p_val_biased = helpers.computeDemographicParity(labels_test_biased, gender_test)

print('Demographic Parity (Blind):', dp_blind)
print('Female proportion (Blind):', p_val_blind, '\n')
# -----
print('Demographic Parity (Biased):', dp_biased)
print('Female proportion (Biased):', p_val_biased)
```

```
EO_blind, n_selecteed_blind = helpers.computeEqualityOfOpportunity(labels_test_blind, gender_test, labels_train_blind, labels_test_blind)

EO_biased, n_selecteed_biased = helpers.computeEqualityOfOpportunity(labels_test_biased, gender_test, labels_train_blind, labels_test_blind)

print('Equal Opportunity (Blind)', EO_blind)
print('Candidates selected (Blind)', n_selecteed_blind, '\n')


print('Equal Opportunity (Biased)', EO_biased)
print('Candidates selected (Biased)', n_selecteed_biased)
```

```
import numpy as np

NDCG_blind = helpers.computeNDCG(labels_test_blind, labels_test_blind)
NDCG_biased = helpers.computeNDCG(labels_test_biased, labels_test_blind)

print("NDCG (Blind):", NDCG_blind)
print("NDCG (Biased):", NDCG_biased)

```

```
print("-----KL DIVERGENCE-----")
print('Blind:', blind_KL)
print('Mean male (Blind):', blind_mean_male)
print('Mean female (Blind):', blind_mean_female, "\n")

print('Biased:', biased_KL)
print('Mean male (Biased):', biased_mean_male)
print('Mean female (Biased):', biased_mean_female, "\n")

print("-----DEMOGRAPHIC PARITY-----")
print('Demographic Parity (Blind):', dp_blind)
print('Female proportion (Blind):', p_val_blind, '\n')

print('Demographic Parity (Biased):', dp_biased)
print('Female proportion (Biased):', p_val_biased, '\n')

print("-----EQUAL OPPORTUNITY-----")
print('Equal Opportunity (Blind)', EO_blind)
print('Candidates selected (Blind)', n_selecteed_blind, '\n')


print('Equal Opportunity (Biased)', EO_biased)
print('Candidates selected (Biased)', n_selecteed_biased, '\n')

print("-----EQUAL OPPORTUNITY-----")
print("NDCG (Blind):", NDCG_blind)
print("NDCG (Biased):", NDCG_biased)
```
